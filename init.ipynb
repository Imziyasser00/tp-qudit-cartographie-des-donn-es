```python
r = pd.read_csv("data/results.csv")
g = pd.read_csv("data/goalscorers.csv")
s = pd.read_csv("data/shootouts.csv")
key = ["date", "home_team", "away_team"]

print("Doublons dans results :", r.duplicated(subset=key).sum())
print("Buts sans match :", (~g.merge(r[key], on=key, how="left", indicator=True)["_merge"].eq("both")).sum())
print("Tirs au but sans match :", (~s.merge(r[key], on=key, how="left", indicator=True)["_merge"].eq("both")).sum())
```
